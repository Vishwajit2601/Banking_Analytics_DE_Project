### Create External Locations

```python
%sql 

-- Raw container location 
CREATE EXTERNAL LOCATION raw_location
URL 'abfss://raw@adlsbankinganalytics.dfs.core.windows.net/' 
WITH (STORAGE CREDENTIAL storage_credential_banking); 

-- Bronze container location 
CREATE EXTERNAL LOCATION bronze_location 
URL 'abfss://bronze@adlsbankinganalytics.dfs.core.windows.net/' 
WITH (STORAGE CREDENTIAL storage_credential_banking); 

-- Silver container location 
CREATE EXTERNAL LOCATION silver_location
URL 'abfss://silver@adlsbankinganalytics.dfs.core.windows.net/' 
WITH (STORAGE CREDENTIAL storage_credential_banking); 

-- Gold container location 
CREATE EXTERNAL LOCATION gold_location
URL 'abfss://gold@adlsbankinganalytics.dfs.core.windows.net/' 
WITH (STORAGE CREDENTIAL storage_credential_banking); 

-- Show all external locations 
SHOW EXTERNAL LOCATIONS;
```

### Create Catalog & Schemas

```python
%sql 
-- Create banking catalog 
CREATE CATALOG banking_catalog 
COMMENT "Banking analytics catalog" 
MANAGED LOCATION 'abfss://bronze@adlsbankinganalytics.dfs.core.windows.net/'; 

-- Create schemas for medallion layers 
CREATE SCHEMA banking_catalog.bronze 
COMMENT "Raw banking data" 
MANAGED LOCATION 'abfss://bronze@adlsbankinganalytics.dfs.core.windows.net/'; 

CREATE SCHEMA banking_catalog.silver 
COMMENT "Cleaned banking data" 
MANAGED LOCATION 'abfss://silver@adlsbankinganalytics.dfs.core.windows.net/';

CREATE SCHEMA banking_catalog.gold 
COMMENT "Aggregated banking analytics" 
MANAGED LOCATION 'abfss://gold@adlsbankinganalytics.dfs.core.windows.net/';
```

```python
%sql
-- Show catalogs 
SHOW CATALOGS; 
```
